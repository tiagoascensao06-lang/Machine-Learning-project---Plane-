# IMPORT 



```python
import pandas as pd


from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

from sklearn.preprocessing import  StandardScaler

learn_data=pd.read_csv('learn.csv')
```

## LOGISTIC REGRESSION
This initially is just for Time Place and Type of operation



## 1º Seperate the data

```python
##Selecting features
feature_cols=['ev_year','ev_month','ev_dow','ev_hour','apt_dist','far_part','type_fly','flt_plan_filed']
numeric_col=['ev_year', 'ev_hour','ev_month','apt_dist']


X=learn_data[feature_cols]
y=learn_data['fatal']


categorical_features = [
    "ev_dow",
    "far_part",
    "type_fly",
    "flt_plan_filed",
]
```

### split the data

```python
X_train,X_val,y_train,y_val=train_test_split(X,y,train_size=0.2,random_state=40,stratify=y)
```

### Dealing with missing values 

```python
for col in numeric_col:
    median_val=X_train[col].median()
    X_train[col]=X_train[col].fillna(median_val)
    X_val[col]=X_val[col].fillna(median_val)


#Mark the missing categories as unknown 
for col in categorical_features:
    X_train[col]=X_train[col].fillna('Unknown')
    X_val[col]=X_val[col].fillna('Unknown')

X_train=pd.get_dummies(X_train)
X_val=pd.get_dummies(X_val)


X_val = X_val.reindex(columns=X_train.columns, fill_value=0)

```

## SCALLING ON TRAINING DATA THEN APPLYING SCALE TO VALIDATION DATA

```python
scl=StandardScaler()

X_train[numeric_col]=scl.fit_transform(X_train[numeric_col])
X_val[numeric_col]=scl.transform(X_val[numeric_col])


X_train = pd.get_dummies(X_train)
X_val = pd.get_dummies(X_val)
X_val = X_val.reindex(columns=X_train.columns, fill_value=0)

```

## APPLYING THE REGRESSION



```python
lr=LogisticRegression()
lr.fit(X_train,y_train)

lr_pred=lr.predict(X_val)
lr_pred_prob=lr.predict_proba(X_val)
print(lr_pred_prob)
```
