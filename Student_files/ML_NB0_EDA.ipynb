```python
def missing_level(df,low=5,high=20):
    total_rows=len(df)
    missing_count=df.isnull().sum()
    missing_pct=(missing_count/total_rows)*100
    summary= pd.DataFrame({
        'missing_count': missing_count,
        'missing_pct': missing_pct,
        'dtype': df.dtypes,
    })

    conditions = [
        (summary["missing_pct"]==0),
        (summary["missing_pct"]>0) & (summary["missing_pct"] <= low),
        (summary["missing_pct"]> low) & (summary["missing_pct"]<=high),
        (summary["missing_pct"]>high),
    ]

    choices=[
        "Complete (0%)",
        f"Low (<= {low}%)",
        f"Intermediate ({low}% - {high}%)",
        f"High (> {high}%)"

    ]
    summary['missing_level']=np.select(conditions,choices,default='Unknown')
    missing_summary = summary[summary["missing_count"] > 0].sort_values(
        by="missing_pct", ascending=False
    )
    return missing_summary


missing_df = missing_level(learn_data, low=5.0, high=20.0)
print(missing_df.sample(10))

```

**Values with high missing rate require a more delicate approach meanwhile values with low missing value can be dealt with just by computing the median or KNN.On the other hand, intermediate misisng can be solved via Median/mode imputation combined with missing flags.**

## Time patterns
### Hour

Most accidents occur from 13:00 to 23:00, with fatality rates mostly around 13.5%–17.8%. Rates are higher in some early morning hours, but those hours have few accidents, so the estimates are less stable. Nineteen rows have no recorded hour; keep them as missing rather than treating them as a real hour.


### Weekday

Fatality rates are fairly similar across weekdays, ranging from 15.7% to 18.3%. There is no strong weekday pattern in this dataset.

### Month

Fatality rates are lower from March through September (about 14.5%–16.6%) and higher in October through December (about 19.8%–22%). Month may be useful to retain as a feature, but these unadjusted rates do not show that month causes higher severity


### Year

Annual fatality rates range from about 15.5% to 20%, with no clear upward or downward trend. Keep year available for modeling, but avoid claiming a long-term trend from these values alone.
## Place and operation

### Airport distance (`apt_dist`)

About 77% of rows have a distance of zero. Fatality is 14.1% for those rows and 26.8% when distance is positive. Zero appears to be a meaningful “at the airport” value, not a missing value; preserve it and consider adding an indicator for positive distance.

### Country (`ev_country`)

Every row has the same country value. It cannot distinguish accidents in this dataset, so drop it before modeling.

### Regulation (`far_part`)

Part 121 accidents have a 1.0% fatality rate (299 rows), while Part 91 has 17.5% (14,964 rows) and Part 135 has 19.7% (547 rows). Other categories are small, so their rates are unstable. Retain the feature, but handle rare categories carefully.

### Flight purpose (`type_fly`)

The largest groups include personal flights (10,923 rows; 18.8% fatal), instructional flights (2,319; 9.2%), aerial application (843; 12.1%), and aerial observation (232; 22.8%). These differences may help prediction, but rates for smaller categories need caution.

### Flight plan (`flt_plan_filed`)

IFR has a 26.3% fatality rate (1,611 rows), VFR 10.2% (882), and NONE 16.6% (12,752). Keep “unknown” and blank values distinct from “none”; they do not mean that no flight plan was filed.



## 4. Sentinel strings in categorical columns

Missing values are not always empty cells: codes such as `UNK` can hide them. We scan every categorical column for common sentinel terms.

```python
sent_term = ['UNK', 'UNKNOWN', 'N/A', 'MISSING', '?','NR']

sent_summary = []
for col in learn_data.select_dtypes(include=['object', 'category']).columns:
    col_str = learn_data[col].astype(str).str.upper()
    for term in sent_term:
        count = (col_str == term).sum()
        if count > 0:
            sent_summary.append({'Feature': col, 'Sentinel Value': term, 'Count': count})

pd.DataFrame(sent_summary)
```

## 5. Target analysis: fatal

The target is `fatal`: 0 means no fatalities and 1 means at least one fatality. We examine the class balance without changing the data.

```python
# 0 = non-fatal, 1 = fatal
counts = learn_data['fatal'].value_counts().sort_index()
percentages = learn_data['fatal'].value_counts(normalize=True).sort_index() * 100

print('Counts:')
print(counts)
print('Percentages:')
print(percentages.round(2))

counts.plot(kind='bar', color=['steelblue', 'indianred'])
plt.title('Fatal vs non-fatal accidents')
plt.xlabel('Fatal (0 = no, 1 = yes)')
plt.ylabel('Number of accidents')
plt.xticks(rotation=0)
plt.show()
```

**Interpretation:** There are 14,090 non-fatal accidents (82.98%) and 2,889 fatal accidents (17.02%), with no missing target values. The target is imbalanced: non-fatal accidents are about 4.9 times as common. A model that always predicts non-fatal would achieve 82.98% accuracy while detecting no fatal accidents. Later modelling should use a stratified split and assess fatal-class recall, precision and F1 alongside accuracy.
